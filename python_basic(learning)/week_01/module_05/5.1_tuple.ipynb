 #dictionary in python


```python
#empty
dic = {}
print(type(dic))

dic = {"Name":"phitron","age":20,"address":"Dhaka","number":[10,20,30,40]}
print(type(dic),dic)
```

```python
#access
print(dic["Name"])
print(dic.get("age"))

#chance value
dic["Name"] = ["Mariya"]
print(dic)
```

Dectionary method

```python
# adding value

dic = {"Name":"phitron","age":20,"address":"Dhaka","number":[10,20,30,40]}
print(dic)
dic["mathe_marks"] = 30
print(dic)
dic.update({"english_marks":40})
print(dic)

```

```python
#delete value
del dic["mathe_marks"]
print(dic)
```

```python
#dictionary view method
dic = {"Name":"phitron","age":20,"address":"Dhaka","number":[10,20,30,40]}
keys = dic.keys()
print(keys)

dic["mathe_numbers"] = 30
print(keys)

values = dic.values()
print(values)

items = dic.items()
print(items)
```

```python
#iteration
for key,value in dic.items():
    print(key,value)
```

Dictionary Comprehension

```python
square = {x:x**2 for x in range(1,11) if x%2==0}
print(square)
```

```python
co_ordinations = [(10,10.5),(20.5,192),(101,102)]
Location =["dhaka","chattogram","sylhet"]

extra_location = {co_or : loc for co_or,loc in zip(co_ordinations,Location)}
print(extra_location)
```
